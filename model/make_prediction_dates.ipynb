#### Make a dataframe of prediction dates and numbers of days, week and months for use by models with different time windows.  The datetime values for the weeks maybe useful for charts.

```python
import pandas as pd
import numpy as np
```

```python
day1 = pd.to_datetime('1998-01-09')
```

```python
theDays = ['03-31', '06-30', '09-30', '12-31']
```

```python
theYears = list(range(1998, 2019))
```

```python
def makePredDates(days, years):
    ''' 
    days is list of string MO-DD, years is list of int
    return list of strings YY-MM-DD
    '''
    theDates = []
    for year in years:
        for day in days:
            theDates.append(str(year) + '-' + day)
    return theDates
```

```python
theDatesStrs = makePredDates(theDays, theYears)
```

```python
theDates = list(map(pd.to_datetime, theDatesStrs))
```

```python
# theDates[:5], theDates[-5:]
```

```python
dfPredTimes = pd.DataFrame(theDates, columns=['predDate'])
```

```python
# dfPredTimes.head()
```

```python
weeks = dfPredTimes.predDate.apply(lambda x: x.to_period('W'))
```

```python
dfPredTimes = dfPredTimes.assign(predWeek=weeks)
# dfPredTimes.head()
```

```python
months = dfPredTimes.predDate.apply(lambda x: x.to_period('M'))
dfPredTimes = dfPredTimes.assign(predMonth=months)
# dfPredTimes.head()
```

```python
day0 = pd.to_datetime('1998-01-09')
```

```python
num_days = (dfPredTimes.predDate - day0).map(lambda x: x.days)
```

```python
dfPredTimes = dfPredTimes.assign(num_days=num_days)
# dfPredTimes.head()
```

```python
week0 = day0.to_period('W')
```

```python
num_weeks = (dfPredTimes['predWeek'] - week0).map(lambda x: x.n)
```

```python
dfPredTimes = dfPredTimes.assign(num_weeks=num_weeks)
# dfPredTimes.head()
```

```python
month0 = day0.to_period('M')
```

```python
num_months = (dfPredTimes['predMonth'] - month0).map(lambda x: x.n)
```

```python
dfPredTimes = dfPredTimes.assign(num_months=num_months)
print(dfPredTimes.head())
print(dfPredTimes.tail())
```

```python
dfPredTimes.to_csv('prediction_times.csv')
```
